```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_rate = df["trend_direction"].str.lower().eq("down").mean()
print(f"Total pages: {df.shape[0]}")
print(f"Declining rate: {declining_rate:.1%}")


stale_and_visible = ((df["days_since_last_update"] >= 180) &
                      (df["impressions_90d"] >= 500)).sum()
print(f"Stale + visible pages (candidates for review): {stale_and_visible}")

print("Starter pipeline Precision@50 — baseline rules: 0.240, random forest: 0.740")
```

## Careful Words — What I Can and Can't Claim

**What my work CAN say:
- Observed:Based on the starter dataset, a meaningful number of pages are both
  stale (180+ days since update) and still visible (500+ impressions) — these are
  real, measured facts about the current data.
- Directional: A learned model (decision tree / random forest) ranks pages more
  precisely than a simple hand-written rule on this data, suggesting there's real,
  learnable signal beyond basic thresholds.
- Decision-support:This ranking can help a content editor prioritize which pages
  to review first, given limited time — it's a tool to guide human judgment, not
  replace it.

What my work will NEVER say:
- No causal proof: I cannot claim that refreshing a page *causes* traffic recovery.
  That would require a controlled experiment (e.g., A/B testing refreshed vs.
  non-refreshed pages), which this observational data cannot provide.
- I am not claiming to have discovered or predicted any
  Google search algorithm factor. I'm only using observable signals (impressions,
  clicks, position, engagement) that were already measured — not reverse-engineering
  how search rankings work.
 A high-priority ranking means a page is "worth reviewing" it
